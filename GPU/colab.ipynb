```python
import json
import pandas as pd

labels = ["Read", "Write", "Execute", "Destructive", "Financial", "Other"]

with open("slm_metrics.json") as file:
      metrics = json.load(file)

matrix = pd.DataFrame(
      metrics["confusion_matrix"],
      index=labels,
      columns=labels
  )

matrix.to_csv("slm_confusion_matrix.csv")
matrix

```
